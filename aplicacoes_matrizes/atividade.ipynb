```python
# Importações

import matplotlib.pyplot as plt
import numpy as np

from scipy import ndimage
from skimage import feature
from skimage import io
```

```python
# Carregando imagem

imagem = io.imread("raiox.png", as_gray=True)
```

```python
# Mostrando imagem original

plt.imshow(imagem, cmap='gray')
plt.title("Imagem Original")
plt.axis('off')
plt.show()
```

```python
# Aplicando laplace

xray_image_laplace_gaussian = ndimage.gaussian_laplace(
    imagem,
    sigma=1
)

plt.imshow(xray_image_laplace_gaussian, cmap="gray")
plt.title("Laplace-Gaussian")
plt.axis("off")
plt.show()
```

```python
# Aplicando Gaussian

xray_image_gaussian_gradient = ndimage.gaussian_gradient_magnitude(
    imagem,
    sigma=2
)

plt.imshow(xray_image_gaussian_gradient, cmap="gray")
plt.title("Gaussian Gradient")
plt.axis("off")
plt.show()
```

```python
# Aplicando Sobel

x_sobel = ndimage.sobel(imagem, axis=0)
y_sobel = ndimage.sobel(imagem, axis=1)

xray_image_sobel = np.hypot(x_sobel, y_sobel)

xray_image_sobel *= 255.0 / np.max(xray_image_sobel)

xray_image_sobel = xray_image_sobel.astype("float32")

plt.imshow(xray_image_sobel, cmap="gray")
plt.title("Sobel")
plt.axis("off")
plt.show()
```

```python
# Aplicando Canny/Prewitt

fourier_gaussian = ndimage.fourier_gaussian(
    imagem,
    sigma=0.05
)

x_prewitt = ndimage.prewitt(
    fourier_gaussian,
    axis=0
)

y_prewitt = ndimage.prewitt(
    fourier_gaussian,
    axis=1
)

xray_image_canny = np.hypot(
    x_prewitt,
    y_prewitt
)

xray_image_canny *= 255.0 / np.max(xray_image_canny)

plt.imshow(xray_image_canny, cmap="gray")
plt.title("Canny/Prewitt")
plt.axis("off")
plt.show()
```
